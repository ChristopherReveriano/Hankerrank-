```python
def parse_line(line):
    parts = line.strip().split()

    station_id = int(parts[0])
    long_w = int(parts[-1])
    lat_n = int(parts[-2])
    state = parts[-3]
    city = " ".join(parts[1:-3])

    return (station_id, city, state, lat_n, long_w)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

# Define schema for the UDF return type
schema = StructType([
    StructField('ID', IntegerType(), True),
    StructField('CITY', StringType(), True),
    StructField('STATE', StringType(), True),
    StructField('LAT_N', IntegerType(), True),
    StructField('LONG_W', IntegerType(), True)
])

# Read text file using serverless-compatible method
df_raw = spark.read.text(path)

# Apply parsing function to each row
parse_udf = udf(parse_line, schema)
df = df_raw.select(parse_udf("value").alias("parsed")).select("parsed.*")

# Show result
df.show(truncate=False)

# Print schema
df.show()
```
